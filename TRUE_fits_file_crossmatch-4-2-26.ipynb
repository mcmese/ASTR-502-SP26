```python
import os
import pandas as pd
import shutil

# Paths 
csv_path = "SPOC_ls_TRUE_power_and_period.csv"          
large_dir = "/Users/youngworlds/Mese/ASTR502/lightcurves/SPOC_Pipeline/mastDownload/TESS"   # directory with 18842 files
output_dir = "/Users/youngworlds/Mese/ASTR502/TRUE_subsample_lightcurves"      # new directory for subset

os.makedirs(output_dir, exist_ok=True)

# Load your target filenames
df = pd.read_csv(csv_path)

# Ensure column is clean
target_files = set(df['file'].astype(str).str.strip())

print(f"Number of target files: {len(target_files)}")

all_files_dict = {}

for root, dirs, files in os.walk(large_dir):
    for f in files:
        if f.endswith(".fits"):
            all_files_dict[f] = os.path.join(root, f)

print(f"Total FITS files found: {len(all_files_dict)}")


matched = []
missing = []

for fname in target_files:
    if fname in all_files_dict:
        matched.append(fname)
    else:
        # Try fast cadence version
        alt_name = fname.replace("_lc.fits", "a_fast-lc.fits")
        
        if alt_name in all_files_dict:
            matched.append(alt_name)
        else:
            missing.append(fname)

print(f"Matched: {len(matched)}")
print(f"Missing: {len(missing)}")

for fname in matched:
    src = all_files_dict[fname]   # <-- full path now
    dst = os.path.join(output_dir, fname)
    shutil.copy2(src, dst)


print("Done copying files.")
```
